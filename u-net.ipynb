```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle

def draw_unet():
    fig, ax = plt.subplots(figsize=(12, 6))
    ax.axis("off")

    layers = [
        ("Input\n128x128", 0, 2.5, 1, 2),
        ("Enc1\n64ch", 1.5, 2.2, 1, 1.6),
        ("Enc2\n128ch", 3, 1.8, 1, 1.2),
        ("Enc3\n256ch", 4.5, 1.4, 1, 0.8),
        ("Bottleneck\n512ch", 6, 1.1, 1.2, 0.5),
        ("Dec3\n256ch", 7.8, 1.4, 1, 0.8),
        ("Dec2\n128ch", 9.3, 1.8, 1, 1.2),
        ("Dec1\n64ch", 10.8, 2.2, 1, 1.6),
        ("Output\n128x128", 12.3, 2.5, 1, 2),
    ]

    for text, x, y, w, h in layers:
        ax.add_patch(Rectangle((x, y), w, h, fill=False, linewidth=2))
        ax.text(x + w/2, y + h/2, text, ha="center", va="center")

    # Encoder arrows
    for i in range(4):
        x1 = layers[i][1] + layers[i][3]
        y1 = layers[i][2] + layers[i][4]/2
        x2 = layers[i+1][1]
        y2 = layers[i+1][2] + layers[i+1][4]/2
        ax.annotate("", xy=(x2, y2), xytext=(x1, y1),
                    arrowprops=dict(arrowstyle="->", linewidth=2))

    # Decoder arrows
    for i in range(4, 8):
        x1 = layers[i][1] + layers[i][3]
        y1 = layers[i][2] + layers[i][4]/2
        x2 = layers[i+1][1]
        y2 = layers[i+1][2] + layers[i+1][4]/2
        ax.annotate("", xy=(x2, y2), xytext=(x1, y1),
                    arrowprops=dict(arrowstyle="->", linewidth=2))

    # Skip connections
    skip_pairs = [(1, 7), (2, 6), (3, 5)]
    for a, b in skip_pairs:
        x1 = layers[a][1] + layers[a][3]/2
        y1 = layers[a][2] + layers[a][4]
        x2 = layers[b][1] + layers[b][3]/2
        y2 = layers[b][2] + layers[b][4]
        ax.annotate("skip / concat", xy=(x2, y2), xytext=(x1, y1),
                    arrowprops=dict(arrowstyle="->", linewidth=1.5),
                    ha="center")

    ax.set_xlim(-0.5, 14)
    ax.set_ylim(0.5, 5.2)
    plt.show()

draw_unet()
```
